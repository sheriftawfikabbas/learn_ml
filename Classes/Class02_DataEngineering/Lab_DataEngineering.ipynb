## The dataset

This is the **Breast Cancer Wisconsin (Diagnostic)** dataset (WDBC, from the
[UCI Machine Learning Repository](https://archive.ics.uci.edu/dataset/17/breast+cancer+wisconsin+diagnostic)).
Each row describes one cell-nucleus image (a digitised fine-needle aspirate of
a breast mass). The columns are:

- `id`: sample identifier (use it as the DataFrame index)
- `diagnosis`: the label — `M` = malignant, `B` = benign
- 30 numeric features, computed per nucleus: `radius`, `texture`, `perimeter`,
  `area`, `smoothness`, `compactness`, `concavity`, `concave points`,
  `symmetry` and `fractal_dimension`, each in three versions — `_mean`
  (average over the nucleus), `_se` (standard error) and `_worst` (mean of the
  three largest values).

This copy has been **deliberately corrupted for teaching**: a few cells contain
the spurious text token `'A'`, some cells are empty, and some column names
break the lowercase convention of the rest. Expect 569 rows and 32 columns
(counting `id`).

## Wrap-up

Cleaning checklist you just practised — and which applies to almost every
tabular dataset:

1. load with the right header/index (`pd.read_csv(..., index_col=...)`);
2. normalise column names;
3. coerce non-numeric junk to `NaN`;
4. impute missing values (median/mean/constant — know the trade-offs);
5. remove or fix sentinel values (zeros, 999s, -1s) using domain knowledge.

Only then start analysing.
