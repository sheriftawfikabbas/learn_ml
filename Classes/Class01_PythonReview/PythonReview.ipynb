## Introduction

Machine learning is about training the machine (a computer program) to make predictions of something. This something could be:
- Is this CT scan that of a COVID-19 patient? Here, a category is predicted: a `yes` or a `no`.
- Does this image indicate cervical cancer? Again, a category is predicted. The machine can also sub-categorise several classes of cancer.
- What will the temperature be tomorrow? Here, a number is predicted.

When you predict a numerical property, the machine learning problem is called a *regression problem*. When you predict a category, it is a *classification problem*.

In both problems, one needs to process *data* using a programming language. This data comes in many possible forms:
- Text data: sequences of characters and words
- Numerical data
- Images: 2D numerical data
- Genomes: DNA sequences, which are essentially text data

Python has evolved into one of the programming languages with the richest set of tools for dealing with these data forms.

## Basic python

Python is quite a nice and easy language to learn and understand. In fact, it is one of the easiest languages to get you **hit the ground running**: you can run your python code, or *script*, virtually anywhere — on your laptop, or even on your phone.

### The `print` function

Let's start with the simplest thing you can do in a program: asking the computer to print something, such as `"Like coffee?"`.

For that we use the `print()` *function*. A function receives some input and performs a task; `print` receives any value and displays it on the screen. The text we hand it below is called a *string* — we will deal with strings properly in a moment.

````python
print("Like coffee?")
````

Notice that the line above looks different from normal text on this page: it is an *executable* code cell, and running the cell executes the code.

The `print` function can also receive several values at once, separated by commas. It prints them one after the other on the same line, separated by spaces:

````python
print("Like coffee?", "Yes, with milk please!")
````

### Strings

A string is any collection of characters — letters, numbers, punctuation, symbols — enclosed between two matching quotation marks. Python accepts **either** single quotes `'...'` **or** double quotes `"..."`; the two forms are equivalent, so `"a"` and `'a'` are the same string. (Using double quotes for strings that themselves contain an apostrophe, like `"Don't panic"`, avoids confusion.)

The string is our first example of a python *data type*: a kind of value the language knows how to work with.

#### f-strings: building strings from variables

Python can insert values into a string using an **f-string**: prefix the string with `f` and put any expression inside curly braces `{}`. Python evaluates the expression and places its result into the string:

````python
element = "silicon"
symbol = "Si"
print(f"The symbol for {element} is {symbol}.")

a = 2
b = 5
print(f"{a} times {b} equals {a * b}.")
````

f-strings are the modern (Python >= 3.6) and most convenient way to format output; you will see them throughout this course.

### Numbers and booleans

Numbers come in several flavours:
- **Integers** (`int`): whole numbers like `5` or `-3`.
- **Floating-point numbers** (`float`): numbers with a fractional part like `3.5` or `2.0`. Python creates a float whenever a decimal point is involved — even `2.0` is a float, not an int.
- **Complex numbers** (`complex`): written with a `j` for the imaginary unit, e.g. `1 + 2j`. These are essential in physics (wavefunctions, impedances) but we will not need them in this course.

The third basic data type is the **boolean** (`bool`): a value that is either `True` or `False`. Booleans are the natural "answer to a yes/no question", and they are what comparison operators produce, as we will see below. Note the capitalisation: python only understands `True` and `False`, not `true` or `TRUE`.

### Variables

A variable is a *name* that refers to a value stored in memory. The `=` operator creates the association: `s = "I love coffee"` means "from now on, the name `s` refers to the string `"I love coffee"`". Reading the variable later retrieves the value; assigning again *rebinds* the name to a new value.

Variable names may contain letters, digits and underscores, but **cannot start with a digit**, are **case-sensitive** (`gap` and `Gap` are different variables), and cannot coincide with python *keywords* (`for`, `if`, `def`, ...). Meaningful names make code vastly easier to read: compare `x = t * 27.2114` with `energy_ev = energy_hartree * 27.2114`.

````python
s = "I love coffee"
print(s)

ready = True
print(ready, type(ready))
````

### Arithmetic operators: `+`, `-`, `*`, `/`, `%`, `//`

Python supports the standard maths operators `+`, `-`, `*` and `/`. Two more deserve an introduction:

- `//` is **floor division**: it divides and then discards the fractional part, so `9 // 4` gives `2`.
- `%` is the **modulus** operator: it gives the *remainder* of the floor division. Since `9 = 2 * 4 + 1`, we have `9 % 4` equal to `1`.

Note that `/` **always returns a float**, even when the division is exact (`8 / 2` is `4.0`). If you want an integer quotient, use `//`.

````python
a = 7
b = 2

print(a + b)    # 9
print(a / b)    # 3.5  (a float)
print(a // b)   # 3    (floor division)
print(a % b)    # 1    (remainder: 7 = 3 * 2 + 1)
print(a ** b)   # 49   (** is exponentiation)
````

### Comparison operators: `==`, `!=`, `<`, `>`, `<=`, `>=`

While the arithmetic operators produce numbers, the comparison operators take two values and produce a **boolean** — they answer questions:

- `a == b`: is `a` exactly equal to `b`?
- `a != b`: is `a` **not** equal to `b`?
- `a < b`: is `a` less than `b`?
- `a > b`: is `a` greater than `b`?
- `a <= b`: is `a` less than **or equal to** `b`?
- `a >= b`: is `a` greater than **or equal to** `b`?

Be careful: equality uses a *double* equals sign `==`. A single `=` is assignment, which is a very common source of bugs in `if` statements.

````python
# Comparison operators produce boolean values
a = 3
b = 5

c = a == b
print(c)
d = a > b
print(d)
e = a <= b
print(e)
````

### Lists

A `list` in python is exactly what its name suggests: an *ordered* list of things — numbers, strings, or even a mix of both. To create a list, enclose the elements in **square brackets** and separate them with commas:
- A list of numbers: `a = [4, 6, 7, 1, 0]`
- A list of strings: `a = ["a", "?", "neptune is a planet"]`
- A mixed list: `a = [3, 0, "Where is my car?"]`

You can also create a list of lists — and nest as deeply as you want: `a = [[1, 2], [3, 4], [5, 6]]` is a list of three elements, each of which is itself a list of two elements.

#### Accessing and changing list elements

Each element occupies a position, its **index**, and python indices **start at 0**: the first element is `a[0]`, the second is `a[1]`, and so on. (Negative indices count from the end: `a[-1]` is the last element.)

````python
a = ["I", "want", "to", "order", "the", 4, "dollars", "mocha"]
print(a[0])
print(a[-1])
````

Because a list is *mutable*, we can change any element by assigning to its slot. We can also ask whether a value belongs to a list with the `in` keyword, which returns a boolean:

````python
a = ["I", "want", "to", "order", "the", 4, "dollars", "mocha"]
a[5] = "yes you can!"
print(a)

print(2 in a)
print("mocha" in a)
````

#### Adding elements to a list

There are three common ways to add elements to a list:
- The `+` operator **creates a new list**: `[1, 2, 3] + [4]` gives `[1, 2, 3, 4]`, leaving the original untouched.
- The `append()` method adds one element to the **end**, modifying the list in place.
- The `insert()` method adds one element at a **chosen index**, also in place.

A classic beginner trap: `append()` and `insert()` modify the list but return `None`, so writing `a = a.append(4)` accidentally erases your list!

````python
a = [1, 2, 3]
print(a + [4])   # a new list; a itself is unchanged
print(a)

a.append(4)      # in place: a is modified
print(a)

a.insert(1, 99)  # insert 99 at index 1
print(a)

print(len(a))    # len() gives the number of elements
````

### Tuples

A tuple, like a list, is an ordered collection — but the elements are enclosed in parentheses `(...)`, and there is one fundamental difference: a tuple is **immutable**. Once created, its elements cannot be changed:

````python
a = (3, 4, 5)
try:
    a[0] = 2
except TypeError as e:
    print("Tuples are immutable:", e)
````

(The `try`/`except` construct catches the error so our notebook keeps running — you will meet it properly later.)

Why have immutable collections at all? Because immutability is a *guarantee*: code that receives a tuple knows its contents cannot be quietly changed elsewhere. Tuples are therefore used for values that should never change (like the physical constants of an element), and they can be used as dictionary keys — lists cannot.

### Dictionaries

In lists and tuples, elements are indexed by integers starting from `0`. A dictionary generalises the idea: it is a collection of key–value pairs `{key: value}`, where each `key` can be any *immutable* value — strings, ints, floats, or tuples — and each `value` can be anything at all. Looking up a key returns the associated value. Keys are unique: assigning to an existing key overwrites its value.

````python
prices = {'apple': 3.5, 'pear': 2.5, 'banana': 4}
print(prices['apple'])

collections = {'a': 'lists', 'b': 'tuples', 'c': 'dictionaries'}
print(collections['c'])

prices['mocha'] = 4.5   # adding a new key
print(prices)
````

### Comments in python

You can ask python to ignore text by putting `#` before it; everything from `#` to the end of the line is a *comment*:
```python
# This line is ignored by python
```
Comments are for humans: use them to explain *why* the code does something, whenever that isn't obvious from the code itself.

### The conditional statement

So far, each statement performed a single unconditional task. Serious programming starts when the computer tests a condition and *then* decides what to do. The `if` statement does exactly that:

```python
if boolean_expression:
    some_statements
```

The indented block runs only if `boolean_expression` evaluates to `True`. Optional `elif` ("else if") and `else` clauses cover further cases. Note python's use of **indentation** to mark the block — it is part of the syntax, not just style.

````python
name = 'Ahmed'
if name[0] == 'A':
    print(name)

temperature = 30
if temperature < 15:
    print('I am wearing a jacket')
elif temperature < 20:
    print('I am wearing a jumper')
else:
    print('I am wearing a t-shirt')
````

### The loop statements

Loops repetitively run a block of statements. They are the starting point of serious programming: most of data science is, at its core, "do this to every element of a big collection".

#### The `for` loop

```python
for x in collection:
    statement1
    statement2
```

`x` is called the **loop variable**: at each iteration it takes the value of the next element of the `collection`, which can be any of the collections we covered (`list`, `tuple`, `dict`) — or a string, which iterates character by character.

````python
for word in ["I", "want", "mocha"]:
    print(word)

prices = {'apple': 3.5, 'mocha': 4.5}
for fruit in prices:
    print(fruit, prices[fruit])
````

#### Repeating N times: `range()`

Very often you don't want to iterate over data — you just want to repeat something `n` times, or generate a sequence of indices. That is what `range()` is for: `range(n)` produces the integers `0, 1, ..., n-1` (like list indexing, it starts at 0 and *excludes* the endpoint):

````python
for i in range(5):
    print(i)

print(list(range(2, 10, 2)))   # start, stop (excluded), step
````

#### The `while` loop

A `while` loop repeats its block *as long as* its condition stays `True`:

```python
while boolean_expression:
    statement1
    statement2
```

Each iteration re-checks the condition. It is your responsibility to make the loop eventually stop — otherwise it runs forever. The compound assignment `a -= 1` is shorthand for `a = a - 1` (and `+=`, `*=`, `/=` work the same way):

````python
a = 10
while a > 0:
    print(a)
    a -= 1
````

### Functions

The `def` statement defines your own *function*: a named, reusable block of code that takes *parameters* and, usually, *returns* a result. We have already used built-in functions (`print`, `len`, `type`); `def` lets you build your own:

```python
def function_name(parameter1, parameter2):
    # body
    return result
```

Functions are the main way to avoid copy-pasting code. Give them a **docstring** — a string in triple quotes right after the `def` line — documenting what they do.

````python
def greet(name):
    """Return a personalised greeting."""
    return "Hello " + name

print(greet("coffee lover"))


def square(x):
    """Return x multiplied by itself."""
    return x * x

print(square(4))
````

### Importing python libraries

One of python's superpowers is its libraries: modules of ready-made functions written by others. You make a library available with `import`. Renaming it with `as` is conventional for the big libraries:

````python
import numpy as np   # numpy: fast numerical arrays — the heart of data python
print(np.log(11))

from numpy import sin   # or import just one function
print(sin(np.pi / 2))
````

`numpy` (numerical python) is the foundation of everything we will do in Class02. `log` and `sin` are examples of its hundreds of mathematical functions.

### Putting it all together

We finish with two tiny science-flavoured programs that combine everything you have seen — variables, functions, collections, loops, conditionals and f-strings.

````python
def celsius_to_kelvin(t_c):
    """Convert a temperature from Celsius to Kelvin."""
    return t_c + 273.15

for t in [0, 25, 100]:
    print(f"{t} °C = {celsius_to_kelvin(t)} K")
````

````python
gap_ev = {'Si': 1.1, 'GaAs': 1.4, 'C': 5.5}   # band gaps in eV

def classify_gap(gap):
    """Classify a material from its band gap (in eV)."""
    if gap == 0:
        return 'metal'
    elif gap < 3.0:
        return 'semiconductor'
    else:
        return 'insulator'

for material, gap in gap_ev.items():
    print(f"{material}: {gap} eV -> {classify_gap(gap)}")
````
